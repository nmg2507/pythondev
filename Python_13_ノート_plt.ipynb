### 課題11　散布図と近似曲線のグラフ.ipynb　用

```python
import numpy as np

x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math

import matplotlib.pyplot as plt
import pandas as pd

#近似直線のグラフを用意
p = np.polyfit(x, y, 1)  #近似直線の傾きaと切片bを求める
f = np.poly1d(p)  #上で求められたpでグラフの元オブジェクトを生成してる
#散布図と近似曲線を描く
plt.scatter(x,y, color="blue", label="score") #散布図
plt.plot(x, f(x), color="red", label="lsm") #近似直線
plt.xlabel("english")
plt.ylabel("math")
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```
