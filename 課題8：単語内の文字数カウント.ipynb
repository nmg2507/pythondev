```python
#課題8 単語内の文字数カウント

word_list = []
alphabet_count = dict.fromkeys(list("abcdefghijklmnopqrstuvwxyz"), 0)  #手打ちりストから辞書作成　.fromkeysで初期値 0

while True:
    word = input ("英単語を入力してください")
    if word == "":
        break
    word_list.append(word)      # Word_list へ入力された単語を入れていく
    for i in list(word):         #word の文字をリスト化して、一文字ずつ i への代入を繰り返す
        alphabet_count[i] +=1

word_list.sort()                            #ワードリストをソートするメソッド
print (f"入力した英単語＝{word_list}")

for key, value in alphabet_count.items():   #辞書の要素数分繰り返す
    if value > 0:
        print (f"{key}が{value}個ありました")

```
