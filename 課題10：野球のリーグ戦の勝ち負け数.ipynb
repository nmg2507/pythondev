```python
#課題　課題10：野球のリーグ戦の勝ち負け数
class Baseball_Team:
    def __init__(self,name,win,lose,draw): #メンバ変数の初期化と設定
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

    def calc_win_rate(self):    #メソッドで勝率を算出
        winningrate = self.win / (self.win + self.lose) 
        return winningrate
    
    def show_team_result(self):    #＋書式設定＆スペース調整
        print(f"{self.name:8s} {self.win:3d}  {self.lose:3d}  {self.draw:3d} {self.calc_win_rate():.3f}")

# インスタンスはリストでつくる　　（あってる？）
teams = [
    Baseball_Team ("Archers", 77, 64, 2),
    Baseball_Team ("Meteors", 71, 69, 3),
    Baseball_Team ("Cheetahs", 69, 68, 6),
    Baseball_Team ("Sharks", 70, 70, 3),
    Baseball_Team ("Pythons", 68, 73, 2),
    Baseball_Team ("Owls", 59, 82, 2)
    ]

# 表の見出し　（べたうちでいい？）
print("team     win lose draw rate") 

# 繰り返しで表示させる
for i in teams:
    i.show_team_result()
```
